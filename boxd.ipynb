## Genre tally from all logged titles

```python
# %load genre_count.py
import pandas as pd

df = pd.read_csv('movies_with_genres.csv')

genre_counts = df['genre'].str.split(', ').explode().value_counts()
print(genre_counts.to_string())
```

## Titles 90 minutes and under

```python
df = pd.read_csv('movies_with_genres.csv')
df[df['runtime'] <= 90].sort_values('year')
```

## Longest runtime

```python
df = pd.read_csv('movies_with_genres.csv')

print(df[df.runtime == df.runtime.max()])
```

## Shortest runtime

```python
print(df[df.runtime == df.runtime.min()])
```

## Total logged runtime

```python
total_minutes = df.runtime.sum()
print(f"{total_minutes/1440} days")
```
