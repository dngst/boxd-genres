# Liked Movies

## Genres

```python
import pandas as pd

df = pd.read_csv('liked_movies_with_genres.csv')

df['genre'].str.split(', ').explode().value_counts()
```

## Longest runtime

```python
df.nlargest(5, 'runtime')
```

## Shortest runtime

```python
df.nsmallest(5, 'runtime')
```

## Total logged runtime

```python
total_minutes = df.runtime.sum()
print(f"{total_minutes/1440} days")
```
